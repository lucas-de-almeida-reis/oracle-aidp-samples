# DB2 Connector Samples

Read/write ingestion samples for the `DB2` connector. External-catalog support is not included in the 4.1 release.


## Ingestion Read

Replace placeholders with values from your environment before running the notebook.


```python
db2_df = spark.read.format("aidataplatform") \
    .option("type", "DB2") \
    .option("host", "<HOST>") \
    .option("port", "<PORT>") \
    .option("database.name", "<DATABASE_NAME>") \
    .option("user.name", "<USERNAME>") \
    .option("password", "<PASSWORD>") \
    .option("schema", "<SCHEMA>") \
    .option("table", "<TABLE_NAME>") \
    .load()

db2_df.show()
```

## Ingestion Write

`CREATE`, `APPEND`, `OVERWRITE`, and `MERGE` are supported. Use `write.merge.keys` with `MERGE`.


```python
db2_df.write.format("aidataplatform") \
    .option("type", "DB2") \
    .option("host", "<HOST>") \
    .option("port", "<PORT>") \
    .option("database.name", "<DATABASE_NAME>") \
    .option("user.name", "<USERNAME>") \
    .option("password", "<PASSWORD>") \
    .option("schema", "<SCHEMA>") \
    .option("table", "<TARGET_TABLE_NAME>") \
    .option("write.mode", "CREATE") \
    .save()

# For MERGE, also add: .option("write.merge.keys", "<KEY_COLUMN>")
```

## Pushdown SQL


```python
db2_pushdown_df = spark.read.format("aidataplatform") \
    .option("type", "DB2") \
    .option("host", "<HOST>") \
    .option("port", "<PORT>") \
    .option("database.name", "<DATABASE_NAME>") \
    .option("user.name", "<USERNAME>") \
    .option("password", "<PASSWORD>") \
    .option("pushdown.sql", "SELECT * FROM <SCHEMA>.<TABLE_NAME> FETCH FIRST 10 ROWS ONLY") \
    .load()

db2_pushdown_df.show()
```

## Connector Options

| Parameter name | Valid values | Mandatory | Description |
| --- | --- | --- | --- |
| `type` | `DB2` | Y | Data source type. |
| `host` | Valid hostname or IP | Y | DB2 host. |
| `port` | Valid port number | Y | DB2 listener port. |
| `database.name` | Valid database name | Y | DB2 database name. |
| `user.name` | Valid DB2 user | Y | Login user used for connectivity. |
| `password` | Valid DB2 password | Y | Password for `user.name`. |
| `schema` | Valid schema name | Y | Schema to read from or write to. |
| `table` | Valid table name | Y for table-based access | Table name inside the schema. |
| `view` | Valid view name | N | View name inside the schema when reading through a view. |
| `pushdown.sql` | SQL query | N | Full SQL pushdown query for reads. |
| `fetch.size`, `row.limit` | Positive integer | N | Read batching and limiting. |
| `partition.column`, `partition.num`, `partition.lower`, `partition.upper` | Partition settings | N | Range-partitioned reads or writes. |
| `pushdown.columns`, `pushdown.filter` | Valid projection or predicate | N | Push work down to DB2. |
| `write.mode` | `CREATE`, `APPEND`, `OVERWRITE`, `MERGE` | Y during write | Write mode to use. |
| `write.empty.value.as.null`, `write.batch.size`, `write.reject.limit` | Boolean or non-negative integer | N | Write controls. |
| `write.merge.keys` | Comma-separated column names | Y for `MERGE` | Key columns used in the `MERGE` condition. |
| `login.timeout.seconds` | Integer from `1` to `300` | N | JDBC login timeout. |
